## Model Training & Comparison
We train four widely-used classifiers with sensible class-imbalance
handling and compare them on Accuracy, Precision, Recall, F1 and ROC-AUC**. 
For a churn use case, Recall (catching as many true churners as possible) and
ROC-AUC (overall separability) matter most — missing a churner is usually more
costly than a false alarm, since a false alarm just means an unnecessary retention offer.

```python
# MODEL TRAINING & COMPARISON

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.base import clone

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

import xgboost as xgb
import pandas as pd

RANDOM_STATE = 42



# TRAIN / TEST SPLIT


X_train, X_test, Y_train, Y_test = train_test_split(x,y,test_size=0.20,random_state=RANDOM_STATE,stratify=y)


# PREPROCESSING

# Scale only the original numerical columns.
# All categorical columns have already been converted to dummy
# variables earlier in your notebook.

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols)
    ],
    remainder='passthrough'
)


# HANDLE CLASS IMBALANCE FOR XGBOOST

negative_count = (Y_train == 0).sum()
positive_count = (Y_train == 1).sum()



# DEFINE MODELS
models = {
    'Logistic Regression': LogisticRegression(
        max_iter=2000,
        class_weight='balanced',
        random_state=RANDOM_STATE
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=300,
        class_weight='balanced',
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    'XGBoost': xgb.XGBClassifier(
        n_estimators=300,
        eval_metric='logloss',
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    'Gradient Boosting': GradientBoostingClassifier(
        n_estimators=200,
        random_state=RANDOM_STATE
    )
}


#  TRAIN AND EVALUATE MODELS
results = []
fitted_pipelines = {}

for name, clf in models.items():

    # Create pipeline
    pipe = Pipeline([
        ('prep', clone(preprocessor)),
        ('clf', clf)
    ])

    # Train model
    pipe.fit(X_train, Y_train)

    # Predictions
    preds = pipe.predict(X_test)

    # Prediction probabilities
    proba = pipe.predict_proba(X_test)[:, 1]

    # Store evaluation results
    results.append({

        'Model': name,
        'Accuracy': accuracy_score(
            Y_test,
            preds
        ),

        'Precision': precision_score(
            Y_test,
            preds,
            zero_division=0
        ),

        'Recall': recall_score(
            Y_test,
            preds,
            zero_division=0
        ),

        'F1-Score': f1_score(
            Y_test,
            preds,
            zero_division=0
        ),

        'ROC-AUC': roc_auc_score(
            Y_test,
            proba
        )
    })

    # Save trained model
    fitted_pipelines[name] = pipe



#  CREATE MODEL COMPARISON TABLE


results_df = (
    pd.DataFrame(results)
    .sort_values(
        'ROC-AUC',
        ascending=False
    )
    .reset_index(drop=True)
)



#  DISPLAY RESULTS


print("MODEL COMPARISON")
print("=" * 70)

print(
    results_df
    .round(4)
    .to_string(index=False)
)



# SELECT BEST MODEL
BEST_MODEL_NAME = results_df.loc[0, 'Model']

best_pipe = fitted_pipelines[BEST_MODEL_NAME]

print("\nBest Model:", BEST_MODEL_NAME)




results_df.style.background_gradient(
    cmap='Purples',
    subset=[
        'Accuracy',
        'Precision',
        'Recall',
        'F1-Score',
        'ROC-AUC'
    ]
)
```

## Hyperparameter Tuning

```python
param_grid = {
    'clf__n_estimators': [200, 400],
    'clf__max_depth': [8, 12, None],
    'clf__min_samples_split': [2, 5],
    'clf__min_samples_leaf': [1, 2],
}

rf_pipe = Pipeline([
    ('prep', preprocessor),
    ('clf', RandomForestClassifier(class_weight='balanced', random_state=RANDOM_STATE, n_jobs=-1))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
grid_search = GridSearchCV(rf_pipe, param_grid, scoring='roc_auc', cv=cv, n_jobs=-1, verbose=0)
grid_search.fit(X_train, y_train)

print("Best CV ROC-AUC:", round(grid_search.best_score_, 4))
print("Best parameters:", grid_search.best_params_)
```

```python
tuned_model = grid_search.best_estimator_
tuned_preds = tuned_model.predict(X_test)
tuned_proba = tuned_model.predict_proba(X_test)[:, 1]

tuned_metrics = {
    'Accuracy': accuracy_score(y_test, tuned_preds),
    'Precision': precision_score(y_test, tuned_preds),
    'Recall': recall_score(y_test, tuned_preds),
    'F1-Score': f1_score(y_test, tuned_preds),
    'ROC-AUC': roc_auc_score(y_test, tuned_proba),
}
print("Tuned Random Forest — Test Set Performance")
for k, v in tuned_metrics.items():
    print(f"  {k:10s}: {v:.4f}")

# Keep whichever is better on the held-out test set: the tuned RF or the original best model
if tuned_metrics['ROC-AUC'] >= results_df.iloc[0]['ROC-AUC']:
    FINAL_MODEL = tuned_model
    FINAL_MODEL_NAME = 'Random Forest (Tuned)'
else:
    FINAL_MODEL = best_pipe
    FINAL_MODEL_NAME = BEST_MODEL_NAME

print(f"\n>>> FINAL MODEL SELECTED: {FINAL_MODEL_NAME}")
```
